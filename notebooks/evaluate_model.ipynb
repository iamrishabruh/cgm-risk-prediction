```python
import sys
from pathlib import Path

import torch
from torch.utils.data import DataLoader, TensorDataset

_cwd = Path.cwd()
ROOT = _cwd.parent if _cwd.name == "notebooks" else _cwd
sys.path.insert(0, str(ROOT))
sys.path.insert(0, str(ROOT / "src"))

from config.config import Config
from evaluation.evaluate import evaluate_torch_model_on_loader, load_torch_checkpoint
```

```python
X_test = torch.randn(50, Config.NUM_FEATURES)
y_test = torch.randint(0, Config.NUM_CLASSES, (50,))
```

```python
ckpt = Config.CHECKPOINTS / "TabTransformer_repFold1.pt"
if not ckpt.exists():
    print("Train first; missing", ckpt)
else:
    model = load_torch_checkpoint("TabTransformer", 1)
    ds = TensorDataset(X_test, y_test)
    loader = DataLoader(ds, batch_size=8, shuffle=False)
    loss, acc, auc, sens, spec = evaluate_torch_model_on_loader(model, loader, is_graph=False)
    print(f"loss={loss:.4f} acc={acc:.4f} auc={auc:.4f} sens={sens:.4f} spec={spec:.4f}")
```
